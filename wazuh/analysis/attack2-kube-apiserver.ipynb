## Analysis of results

The dominant pattern in this capture is a broad, roughly uniform amplification of existing baseline activity rather than the appearance of a new, concentrated signal. Every verb grows by close to the same factor (get x2.2, update x2.0, watch x2.2), and the same holds for `userAgent`: `kube-probe` (660 to 1320), `kubelet` (217 to 656), `kube-controller-manager` leader-election (296 to 592), `kube-scheduler` leader-election (296 to 592), and `kube-apiserver` itself (210 to 420) all scale up by roughly the same x2 factor, with no new account appearing that was absent in the baseline. The `authorization.reason` breakdown confirms this: every RBAC grant already present in the baseline (`system:public-info-viewer`, `system:kube-controller-manager`, `system:kube-scheduler`, `calico-kube-controllers`, and so on) simply increases in volume, without any new ClusterRoleBinding being exercised.

This uniform x2 inflation is consistent with the nf_conntrack saturation mechanism driving this attack: once the connection tracking table fills, packets are dropped randomly, which breaks existing `watch` connections held by these same clients and forces them to reconnect and resynchronize, typically through a full `list` call rather than resuming an efficient `watch`. This produces more API calls from each client, not fewer, even though the clients themselves are not the ones under attack, which explains why the increase shows up broadly across nearly every existing account rather than as a spike concentrated on one of them.

Within this uniform growth, `update` has the smallest factor of the four verbs (x2.0, against x2.2 for `get` and `watch`), while most of the reconnection/resync traffic described above falls under `get`/`list`/`watch`. These read operations are typically served from the kube-apiserver's in-memory watch cache rather than by querying etcd directly, so this traffic increase does not necessarily translate into a proportional increase in etcd load. This is consistent with a separate measurement from the same attack window, where etcd CPU usage falls rather than rises during the attack, even as kube-apiserver CPU and overall client-facing traffic both increase.

`objectRef.namespace` shows the same pattern: the growth is concentrated in `kube-system` (643 to 1284), `default` (116 to 240), and `kube-node-lease` (88 to 176), the namespaces that host normal control plane chatter, while `attacker-ns` stays almost untouched (0 to 6). This is consistent with the attack targeting the `/healthz` endpoint of the `kube-apiserver` directly: `/healthz` is a non-resource URL with no associated `objectRef`, so the attacker's own requests do not themselves populate the `objectRef.resource`/`objectRef.namespace` breakdown. What we observe instead is the knock-on effect described above: existing clients reconnecting and resyncing more often, which inflates their own normal traffic rather than introducing a distinct new signal.

`patch_on_pods` stays at 0 in both scenarios, confirming that this attack does not touch pod status at all, unlike mechanisms built on the readiness probe. `objectRef.subresource: status` barely moves (3 to 12, a negligible absolute count), reinforcing that the attack's footprint is not expressed through subresource state churn.

`sourceIPs` shows one notable new entry, `192.168.62.165` (0 to 252), alongside the growth of existing source IPs (`10.10.10.10`: 1572 to 3238, `::1`: 210 to 420, `10.10.10.11`: 83 to 276). The loopback address `::1` doubling is itself a side effect worth noting: it reflects the `kube-apiserver`'s own internal health-check traffic increasing under load, not external attacker activity.

## Detectability analysis

**1. No single field stands out sharply.** Unlike a pattern driven by a specific resource or a new controller account, this attack's signature is diffuse: almost every verb, namespace, and userAgent scales by roughly the same factor. A detection rule keyed on any single field (a specific resource, a specific account) would likely miss this attack, since nothing in the baseline composition qualitatively changes, only its volume.

**2. Aggregate rate is the most reliable indicator available.** `events/sec` average grows x3.9 and peak x2.6. A threshold on overall event rate, independent of which field is driving it, is the most direct way to catch this kind of attack in these logs.

**3. A new source IP appearing at volume.** `192.168.62.165` goes from 0 to 252 occurrences and was entirely absent from the baseline. Tracking new source IPs crossing a volume threshold is a usable secondary indicator, though weaker than the account-based indicators available for other attack types, since it depends on IP visibility and stability.

### Detectability limitations to note

- **The attack's own requests are largely invisible in `objectRef` fields.** Because `/healthz` is a non-resource endpoint, the bulk of the malicious traffic does not appear in `objectRef.resource` or `objectRef.namespace` at all. Detection based on these fields would only catch the secondary effect (other clients retrying more), not the attack itself.
- **No new RBAC grant or controller account appears.** Every `authorization.reason` entry present in the attack was already present in the baseline, just at higher volume. A detector relying on new/unseen accounts or roles, effective for other attack types, would not trigger here.
- **The signal is proportionally weaker than the raw event count suggests.** A x2.3 increase in `total_events` is modest compared to the volume spikes seen in attacks that create new objects or drive status churn, which means a threshold tuned to catch those attacks might be set too high to catch this one reliably.
