## Analysis of results

The breakdown by `(resource, verb)` shows a single, clean dependency chain being exercised: `pods` (patch +3484, get +3546) at the source, propagating to `replicasets` (update +5141) and then `deployments` (update +5142). `endpoints` and `endpointslices` stay essentially flat (64 to 62 and 70 to 68 respectively), confirming that only the pod-replicaset-deployment propagation path is active in this capture.

The `objectRef.subresource: status` field shows a sharp shift: from 3 to 13771 occurrences (x4590), reaching roughly 65% of total attack volume. This is consistent with one hundred independently scheduled pods each flipping readiness every second: nearly every event generated by the attack is a state update, not a structural change to the cluster.

The two controller accounts `deployment-controller` and `replicaset-controller` are entirely silent in the baseline (0 events each) and jointly responsible for 10283 events during the attack, roughly 48% of the total volume. This matches the expectation that a readiness flip reaching the API server fans out into two independent reconciliation loops (ReplicaSet status, then Deployment status) for every single pod, which explains why their combined volume is close in magnitude to the `pods patch`/`get` volume generated directly by the kubelet.

`verb: update` alone accounts for 10944 events (up from 710 in the baseline, x15.4), the largest single-verb contributor to the attack, consistent with the two controllers above each performing one `update` per pod per reconciliation cycle. `patch` goes from 3 to 5195 (x1732), matching the kubelet's periodic `pods/status` patches driven by the readiness probe.

`authorization.decision` stays at 100% `allow` (990 to 4531), and the `authorization.reason` breakdown confirms the attack uses only default RBAC grants: the `ClusterRoleBinding` entries for `system:controller:deployment-controller` and `system:controller:replicaset-controller` appear at high volume (0 to 1739 and 0 to 1730) without any new or unusual role being invoked.

## Detectability analysis

**1. Appearance of service accounts absent from the baseline.** `deployment-controller` and `replicaset-controller` are silent in the baseline and jointly generate 10283 events during the attack. A system account with zero baseline activity suddenly producing thousands of events per observation window is a strong, low false positive indicator.

**2. Concentration on `objectRef.subresource: status`.** The jump from 0.1% to roughly 65% of total volume on the `status` subresource is a very sharp signal. A threshold rule on this ratio would catch this attack with a wide margin.

**3. Volume and rate on `update`/`patch` verbs.** `update` alone grows by a factor of 15.4 and `patch` by a factor of 1732. A rule combining verb distribution with absolute event rate would be more robust than relying on total volume alone, since it ties the alert to the specific mechanism (status churn) rather than a generic traffic spike.

### Detectability limitations to note

- **RBAC decision always `allow`**: a detector based only on authorization denials would miss this attack entirely. Detection must rely on volume and verb distribution, not on the legitimacy of the calls.
- **Namespace confinement**: the entire signal sits in `attacker-ns` in this capture. An attacker spreading the same pattern across several namespaces, or inside a namespace already carrying legitimate churn, would require detection to fall back on the subresource and controller-account indicators rather than namespace filtering.
- **Narrow propagation footprint**: because no Service is attached to these pods, the `endpoints`/`endpointslices` indicators stay flat here. A detection rule relying specifically on endpoint/endpointslice churn would miss this pattern, which argues for keeping the `status` subresource ratio and the deployment/replicaset controller accounts as the primary indicators for this mechanism.
