## Analysis of results

The breakdown by `(resource, verb)` reveals two groups of affected objects that correspond precisely to the two data dependencies exploited by the attack (sections 4.4 and 4.5). On one side, `replicasets` (+3119) and `deployments` (+3129) materialize the pod-replicaset-deployment dependency (Fig. 6): every readiness flip on a pod forces the `kube-apiserver` to propagate the change to the parent `ReplicaSet` and then `Deployment`. On the other side, `endpoints` (+2032) and `endpointslices` (+1964) materialize the pod-endpoint-service dependency (Fig. 7), specific to this attack. This observation confirms empirically, with figures to back it up, that the compromise of a single resource (the pod) amplifies its impact across the entire graph of dependent objects, since both dependency chains show up in our logs with volumes of the same order of magnitude (roughly 2000 to 3100 events each), consistent with the fact that they are triggered by the same source event (the readiness flip) and propagate in parallel rather than sequentially.

The shift of `objectRef.subresource: status` from 0.1% to 49% of the total volume is the clearest signature of this reconciliation mechanism: nearly all the noise generated corresponds to state updates, not structural changes to the cluster, which is consistent with an attack that creates or deletes no business object after the initial deployment phase, but instead shakes the `status` field across the entire dependency graph.

The simultaneous appearance of the four controller accounts (`deployment-controller`, `replicaset-controller`, `endpoint-controller`, `endpointslice-controller`), each entirely silent in the baseline, is a direct and unambiguous fingerprint of this dual propagation: their combined volume (10244 events) accounts for 60% of the total generated by the attack on its own, making them the main contributor to the observed signal, more so than the attacker's own requests.

Finally, the fact that `authorization.decision` stays at 100% `allow` throughout the attack confirms that it exploits no authorization flaw: it stays, start to finish, within the default RBAC permissions of a pod's `service account token` (Table 1 of the paper: `pods/status: patch`, `services`/`endpoints: watch`), consistent with the paper's threat model (section 3.1) where the attacker needs no elevated privileges.

## Detectability analysis

**1. Appearance of service accounts absent from the baseline.** A system account that is usually silent and suddenly becomes active is an indicator with a very low false positive rate: binary detection by presence/absence, not a fragile statistical threshold.

**2. Concentration on `objectRef.subresource: status`.** A rule such as "more than X% of events in an interval target the `status` subresource" would be a good candidate for a Wazuh rule, cheap to evaluate.

**3. Volume and rate.** The x8 factor on `total_events` and `events/sec` largely exceeds the background noise measured in the baseline (peaks of 25/sec or less in the baseline versus 177/sec during the attack). A simple threshold on the event rate would detect this attack unambiguously.

### Detectability limitations to note

- **RBAC decision always `allow`**: a rule based solely on authorization denials would be blind to this attack. The detector must rely on *volume*, not on the *legitimacy* of the calls.
- **Namespace confinement**: if the attacker spreads activity across several namespaces or inside a shared legitimate namespace, the `objectRef.namespace` indicator loses its discriminating power. In that case, detection would have to rely solely on indicators 1 and 2, which remain valid regardless of namespace.
- **Dependency on the temporal observation point**: as observed during capture (the gap between `@timestamp` and `requestReceivedTimestamp` under load), a real-time detector based on indexing time can temporarily underestimate the real event rate during load spikes.
