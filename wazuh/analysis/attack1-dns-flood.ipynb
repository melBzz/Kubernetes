## Analysis of results

The overall volume barely moves: total events grow by only 17% (2089 to 2449), and two of the five verbs, `update` and `watch`, actually decrease slightly during the attack. This is consistent with the attack mechanism itself: the flood is a raw UDP socket flood against CoreDNS's DNS service (port 53), sent directly from the attacker pods without going through the `kube-apiserver` at any point, so the malicious traffic itself leaves no trace in these logs by construction.

Nearly all of the modest increase that does appear traces back to a single source that has nothing to do with the attack: `kubectl`, `kubernetes-admin`, and `kubeadm:cluster-admins` each jump from 0 to 180 occurrences, together with `objectRef.subresource: exec` (0 to 106) and `responseStatus.code: 101` (0 to 69, the HTTP code for a protocol upgrade, consistent with an interactive `kubectl exec` session). These four fields describe the same thing from different angles: a `kubectl exec` session run as `kubernetes-admin`, which matches exactly how the measurement pipeline works, `capture-metrics-loop.sh` polls CoreDNS's metrics endpoint by executing into the victim pod. The `pods get` count (0 to 180 in the resource/verb breakdown) lines up with this too, since `kubectl exec` first fetches the target pod object before opening the exec session. `objectRef.namespace: safe-ns` rising from 0 to 183 confirms the same thing from the namespace side, since the victim pod that the measurement script execs into lives in `safe-ns`.

Taken together, these fields account for the large majority of the observed delta on their own. `patch` (+14) and `create` (+4) are small enough in absolute terms to plausibly be ordinary baseline variation rather than attack-driven. `patch_on_pods` stays at 0 in both scenarios, confirming no pod status activity is associated with this attack.

No field in this dataset carries a signal that originates from the DNS flood itself. The attacker pods, their UDP traffic, and CoreDNS's internal rejection mechanism (`MaxConcurrentRejectCount`) are entirely absent from the Kubernetes audit trail, which only sees API server activity.

## Detectability analysis

**This attack is not detectable in Kubernetes audit logs.** Every indicator that moves during the attack capture window traces back to the measurement methodology itself (`kubectl exec` polling through the victim pod), not to the attacker's traffic.

**The near-zero native signal is itself informative for this study.** The contrast between a 17% rise in total audit events and the severe impact measured through other telemetry (CPU, packet loss, `nf_conntrack` saturation) demonstrates concretely that Kubernetes audit logging has a blind spot for attacks that operate below the API server, at the kernel/network level.

### Detectability limitations to note

- **Measurement artifacts can masquerade as signal.** The appearance of `kubectl`, `kubernetes-admin`, and `pods/exec` in this capture is a direct result of how the attack's impact was measured (`kubectl exec`-based polling), not of the attack. Any pipeline that captures audit logs alongside its own measurement tooling needs to account for this before attributing observed deltas to the attack under test.
- **No substitute indicator is available in this log source.** Unlike attacks that manipulate Kubernetes objects directly, this attack produces no usable `objectRef`, `verb`, or `userAgent` signal at all. Detecting it requires a different telemetry source entirely, such as CoreDNS's own metrics (`MaxConcurrentRejectCount`) or `nf_conntrack` occupancy, both already used to measure this attack's impact in this study.
